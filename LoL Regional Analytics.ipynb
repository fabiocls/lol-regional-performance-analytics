```
import pandas as pd
import plotly.express as px
pd.set_option('display.max_columns', None)
```

```
df= pd.read_csv('/content/base_dados_projeto_lol.csv')
df.head()
```

##  Definição do Escopo e Recorte da Análise

Para que a nossa análise exploratória seja estatisticamente válida e taticamente comparável, precisamos isolar o ruído da base de dados original. O ecossistema competitivo global de League of Legends é composto por dezenas de ligas e diferentes divisões de base (Academy/Challengers).

Neste bloco, definimos os nossos alvos de estudo:
1. **Ligas Alvo:** `CBLOL` (Brasil), `LCK` (Coreia do Sul) e `LEC` (Europa), representando três culturas de jogo completamente distintas.
2. **Splits Alvo:** Selecionamos exclusivamente o `Split 2` e o `Summer Split` da temporada de 2022. Este recorte é crucial porque a segunda metade do ano reflete o auge tático das equipas, onde os elencos estão consolidados após a janela de transferências do meio do ano e o foco total é a qualificação para o *World Championship* (Mundial).

```
ligas_alvo = ['CBLOL' , 'LCK' , 'LEC']
splits_alvo = ['Split 2' , 'Summer']
```

###  Filtragem por Sumário de Equipas (`position == 'team'`)

Uma particularidade do formato de dados fornecido pelo *Oracle's Elixir* é que cada mapa gera 12 linhas: 10 para o desempenho individual de cada jogador e 2 linhas que consolidam as estatísticas globais das equipas (Lado Azul e Lado Vermelho).

Para as nossas métricas de macrofutebol (como tempo total, abates totais e placar de visão da equipa), precisamos de usar apenas as linhas consolidadas de equipa. Caso contrário, se usássemos as linhas dos jogadores, estaríamos a inflacionar e a quintuplicar os dados de tempo e abates do mapa. Filtramos isso definindo `position == 'team'` e aplicamos o método `.copy()` para evitar avisos de *SettingWithCopyWarning* no Pandas durante as transformações futuras.

```
df_filtrado = df.query("league in @ligas_alvo and split in @splits_alvo and position == 'team'").copy()
```

### Conversão de Tempo



```
df_filtrado['gamelength'] = df['gamelength'] / 60
```

```
df_filtrado.head()
```

##  Agregação por Partida Única e Resumo de Letalidade

Embora tenhamos isolado as linhas de equipa, cada confronto competitivo ainda ocupa **duas linhas** na tabela `df_filtrado` (uma linha para o sumário do time Blue e outra para o sumário do time Red).

Para analisarmos a letalidade real de um jogo completo, precisamos de agrupar essas duas linhas num registo único por `gameid`. Construímos um dicionário de agregação personalizado no `.groupby()` aplicando regras específicas para cada tipo de dado:
* `kills: 'sum'` ➡️ Soma os abates dos dois lados, gerando o total de mortes da partida.
* `gamelength: 'first'` ➡️ Como o tempo de jogo é idêntica para os dois lados, extraímos o primeiro valor de forma íntegra.
* `teamname: lambda x: ' vs '.join(x)` ➡️  Criamos uma função lambda que une o nome das duas equipas com um conector `" vs "`.

```
km = df_filtrado.groupby('gameid').agg({
    'kills': 'sum',
    'gamelength': 'first',
    'league' : 'first',
    'ckpm' : 'first',
    'teamname': lambda x: ' vs '.join(x),
}).copy()
km.reset_index(inplace = True)
```

```
km
```

##  Análise de Ritmo: Índice de Abates Combinados por Minuto (CKPM)

Com a nossa base consolidada por partida única na tabela `km`, iniciamos a fase de análise visual avaliando o **CKPM** (*Combined Kills Per Minute*). Essa métrica calcula a taxa de letalidade ativa do jogo através da fórmula:

$$CKPM = \frac{\text{Abates Totais da Partida}}{\text{Duração da Partida em Minutos}}$$

**Nota Estatística sobre a Métrica:** Como o CKPM já é uma taxa calculada por partida, cada linha da nossa tabela já guarda o ritmo médio daquele confronto específico. No código abaixo, utilizamos o `px.strip` (gráfico de dispersão unidimensional) para plotar **cada uma das partidas como um ponto individual**. O gráfico não calcula uma média matemática no código; ele expõe a distribuição bruta de todos os jogos, permitindo avaliar visualmente onde os dados se concentram e a frequência de casos extremos (*outliers*).

```
fig = px.strip(
    data_frame=km.reset_index(),
    x='league',
    y='ckpm',
    color='league',
    hover_data=['gameid', 'kills', 'gamelength'],
    title='Distribuição de Cada Jogo por Liga'
)

fig.update_layout(template='plotly_white', xaxis_title='Liga', yaxis_title='CKPM')
fig.show()
```

###  Análise do Gráfico: A Assinatura de Ritmo de Cada Região

A distribuição vertical e a densidade das "nuvens" de pontos revelam assinaturas culturais muito claras para cada uma das regiões estudadas:

* **CBLOL (Nuvem Azul):** Exibe uma forte concentração de pontos deslocada para o topo do gráfico, orbitando de forma densa entre as faixas de **0.8 e 1.2 CKPM**. Isso comprova que o ecossistema brasileiro possui uma cultura de jogo voltada para lutas constantes e agressões contínuas no mapa.
* **LCK (Nuvem Laranja):** Mostra um comportamento inversamente oposto, com seus pontos fortemente concentrados na base inferior do gráfico (faixa de **0.4 a 0.7 CKPM**). Esse desenho quantifica o estilo coreano: um jogo cirúrgico, focado em controle tático, punição de erros e negação sistemática de riscos. Os poucos pontos isolados no topo provam que, quando a estratégia inicial falha, a LCK também experiência partidas caóticas.
* **LEC (Nuvem Verde):** Apresenta uma distribuição vertical muito mais uniforme e espalhada. Isso indica que a Europa não possui uma identidade de ritmo tão homogênea; a letalidade flutua livremente dependendo da preferência individual dos times.

##  Gerenciamento de Tempo: Duração Média das Partidas

Compreendido o ritmo de abates, a próxima pergunta essencial é: *Ligas mais agressivas destroem o Nexus mais rápido?* Para responder a isso, calculamos o tempo médio de duração das partidas (`gamelength`) por liga. No gráfico gerado a seguir, aplicamos um corte de visualização no eixo Y entre **30 e 34 minutos**. Esse "zoom" visual ajuda a evidenciar as sutis diferenças de minutos e segundos que separam o encerramento das partidas em cada ecossistema.

```
tempo_partida = df_filtrado.groupby('league').agg({
    'gamelength': 'mean'
}).copy()
tempo_partida.reset_index(inplace = True)
```

```
tempo_partida
```

```
fig = px.bar(
    data_frame=tempo_partida,
    x='league',
    y='gamelength',
    color='league',
    title='Comparação Direta de Tempo de Jogo por Liga'
)


fig.update_layout(
    xaxis_title='Liga',
    yaxis_title='Tempo Total (Escala da Tabela)',
    template='plotly_white'
)
fig.update_yaxes(range=[30, 34])

fig.show()
```

### Análise do Gráfico: O Mito da Duração de Jogo Desmistificado

A visualização traz uma revelação valiosa: **as partidas profissionais duram quase exatamente o mesmo tempo no mundo inteiro**, orbitando muito próximo da barreira dos **33 minutos**.

* O **CBLOL** é sutilmente a liga mais rápida na média (**32.7 min**), enquanto a **LEC** é a mais demorada (**33.5 min**).
* **O Insight Estratégico:** Cruzando esse resultado com o gráfico anterior de CKPM, chegamos a uma grande conclusão tática. A real divergência entre o LoL coreano (LCK) e o brasileiro (CBLOL) não está no relógio final do jogo. A diferença é sobre **como esses 33 minutos são preenchidos**: enquanto o Brasil gasta o tempo acumulando abates e lutas, a Coreia do Sul gasta os mesmos 33 minutos em rotações meticulosas, controle de visão e cerco a objetivos neutros.

##  A Guerra de Visão: Wards por Minuto (WPM) vs Placar de Visão (VSPM)

Para aprofundar o entendimento sobre o controle estratégico de cada região, isolamos duas métricas fundamentais de mapa: `wpm` (quantidade de sentinelas colocadas por minuto) e `vspm` (placar geral de visão obtido por minuto pelas equipes).

 Geramos um gráfico de barras agrupadas para colocar essas duas métricas lado a lado por liga, permitindo avaliar não apenas o volume de recursos de visão investidos, mas a eficiência real desse investimento.

```
vision_per_min = df_filtrado.groupby('league').agg({
    'wpm': 'mean',
    'vspm' : 'mean',

}).copy()
vision_per_min.reset_index(inplace = True)
```

```
vision_per_min
```

```
fig = px.bar(
    data_frame=vision_per_min,
    x='league',
    y=['wpm', 'vspm'],
    barmode='group',
    title='Guerra de Visão: Wards por Minuto (WPM) vs Placar de Visão por Minuto (VSPM)'
)


fig.update_layout(
    xaxis_title='Liga',
    yaxis_title='Média por Minuto',
    legend_title='Métricas de Visão',
    template='plotly_white'
)

fig.show()
```

### 💡 Análise do Gráfico: Eficiência Tática na Iluminação do Mapa

Os dados revelam um padrão fascinante sobre a eficiência do controle de mapa:

* **O Volume de Investimento:** Note que o **CBLOL (3.50 WPM)** e a **LCK (3.54 WPM)** colocam praticamente a mesma quantidade de sentinelas no chão a cada minuto de jogo.
* **A Diferença na Eficiência:** Embora depositem o mesmo número de wards no mapa, a LCK se descola na liderança com o **VSPM de 7.92** em relação ao CBLOL (7.59). Como o Placar de Visão pontua a negação de wards inimigas e a eficácia em detectar oponentes, esse gráfico prova que os times coreanos extraem muito mais inteligência tática do mesmo número de wards colocadas, posicionando-as nos locais mais cirúrgicos possíveis.

##  Dinâmica de Early Game: Vantagem aos 15 Minutos dos Vencedores

Por fim, investigamos a capacidade de transição e resiliência de cada liga. Filtramos a base de dados para isolar apenas as equipes que **venceram as partidas (`result == 1`)** e avaliamos qual era o saldo de diferença de ouro que elas possuíam na marca exata dos 15 minutos (`golddiffat15`) — ponto que marca o fim da fase de rotas.

Utilizamos um **Boxplot com pontos sobrepostos** para identificar a mediana dessas vantagens e mapear a ocorrência de grandes reviravoltas. A linha horizontal cinza fixada no ponto **ZERO** serve como referência de empate absoluto aos 15 minutos; logo, **qualquer ponto plotado no quadrante negativo representa um "Comeback" (Vira-Vira)**.

```
df_filtrado[['golddiffat15' , 'result' , 'teamname' , 'gameid' , 'league']].head(6)
```

```
vant_15_min = df_filtrado[df_filtrado['result'] == 1] [['golddiffat15' , 'teamname' , 'league' ,  'result' ]].copy().reset_index()
```

```
fig = px.box(
    data_frame=vant_15_min,
    x='league',
    y='golddiffat15',
    color='league',
    points='all',
    hover_data=['teamname'],
    title='Vantagem de Ouro aos 15 Minutos dos Times Vencedores'
)


fig.add_hline(y=0, line_dash="dash", line_color="gray", annotation_text="Jogo Empatado aos 15 min")

fig.update_layout(
    xaxis_title='Liga',
    yaxis_title='Diferença de Ouro aos 15 min (golddiffat15)',
    template='plotly_white'
)

fig.show()
```

### 💡 Análise do Gráfico: Imprevisibilidade Caótica vs Rigidez Macro

O Boxplot expõe perfeitamente o nível de estabilidade macroeconômica de cada ecossistema:

* **CBLOL -  A caixa azul abraça a linha do zero e estende raízes profundas no quadrante negativo. O CBLOL possui uma nuvem densa de partidas abaixo da linha de empate, incluindo um caso extremo (*outlier*) no índice `103594`, onde a equipe da **Miners** conseguiu vencer a partida mesmo enfrentando uma desvantagem colossal de **-5.245 de ouro aos 15 minutos** contra a RED Canids. Isso prova que o início de jogo no Brasil não dita o resultado final; a liga é altamente imprevisível e propensa a reviravoltas.
* **LCK - Em total contraste, a nuvem laranja da Coreia do Sul é extremamente magra abaixo da linha de zero, com quase nenhum ponto ultrapassando a barreira dos -2.000 de ouro. Isso atesta o controle macro da LCK: quando um time coreano fecha a fase de rotas em vantagem financeira, ele joga de forma tão segura e protocolar que o adversário raramente encontra brechas para uma virada. Quem fica rico cedo na LCK, geralmente vence de forma limpa.
* **LEC - A Europa se posiciona como uma zona de transição equilibrada. Suas caixas estão consolidadas acima do zero, indicando previsibilidade, mas a região ainda apresenta uma quantidade moderada de reviravoltas, sem atingir os extremos caóticos vistos no Brasil.
